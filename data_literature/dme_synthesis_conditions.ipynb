# 1 Direct synthesis of dimethyl ether from synthesis gas

Source: *Direct synthesis of dimethyl ether from synthesis gas: Experimental study and mathematical modeling* (Chemical Engineering Journal, 2017, DOI: 10.1016/j.cej.2017.04.132) [web:13].


## Reaction system

The paper studies direct DME synthesis from syngas using a bifunctional catalyst system [web:13].

1. Methanol synthesis: `CO + 2H2 <-> CH3OH`
2. Methanol dehydration: `2CH3OH <-> CH3OCH3 + H2O`
3. Overall: `2CO + 4H2 <-> CH3OCH3 + H2O`

The paper reports thermodynamic analysis over **10–100 bar**, **220–280 °C**, and **H2/CO = 1–6** [web:13].

## Optimal reaction conditions



| Parameter | Value |
|---|---:|
| Pressure | 30 bar |
| Temperature | 280 °C |
| Initial H2/CO ratio | 2:1 |
|  DME  | 24 - 40 % |
|  Methanol  | 6 - 26 % |
