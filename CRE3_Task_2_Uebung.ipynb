## 1  Allgemeine Form der Shomate-Gleichungen

Die reduzierte Temperaturvariable ist definiert als:

$$t = \frac{T \;[\mathrm{K}]}{1000}$$

### 1.1  Wärmekapazität

$$\boxed{C_p^\circ(T) = A + B\,t + C\,t^2 + D\,t^3 + \frac{E}{t^2}}$$

Einheit: $\mathrm{J\,mol^{-1}\,K^{-1}}$

### 1.2  Standardenthalpie (relativ zu 298.15 K)

$$\boxed{H^\circ(T) - H^\circ_{298} = A\,t + \frac{B\,t^2}{2} + \frac{C\,t^3}{3} + \frac{D\,t^4}{4} - \frac{E}{t} + F - H}$$

Einheit: $\mathrm{kJ\,mol^{-1}}$

> **Hinweis:** $H$ (Koeffizient) ≠ $H^\circ_f$. Der Koeffizient $H$ ist der NIST-Referenzwert
> zur Berechnung der Enthalpiedifferenz, **nicht** die Standardbildungsenthalpie.

### 1.3  Standardentropie

$$\boxed{S^\circ(T) = A\,\ln(t) + B\,t + \frac{C\,t^2}{2} + \frac{D\,t^3}{3} - \frac{E}{2\,t^2} + G}$$

Einheit: $\mathrm{J\,mol^{-1}\,K^{-1}}$

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as scp

# ---------------------------------------------------------------------------
# Shomate-Koeffizienten: {Spezies: [(T_min, T_max, A, B, C, D, E, F, G, H), ...]}
# ---------------------------------------------------------------------------
SHOMATE = {
    'CH4': [
        (298, 1300,
         -0.703029,  108.4773, -42.52157,  5.862788,
          0.678565,  -76.84376, 158.7163,  -74.87310)
    ],
    'CO2': [
        (298, 1200,
          24.99735,   55.18696, -33.69137,  7.948387,
          -0.136638, -403.6075,  228.2431, -393.5224)
    ],
    'CO': [
        (298, 1300,
          25.56759,    6.096130,  4.054656, -2.671301,
           0.131021, -118.0089,  227.3665, -110.5271)
    ],
    'H2': [
        (298, 1000,
          33.066178,  -11.363417,  11.432816, -2.772874,
          -0.158558,   -9.980797,  172.707974,  0.000000),
        (1000, 2500,
          18.563083,   12.257357,  -2.859786,  0.268238,
           1.977990,   -1.147438,  156.288133,  0.000000)
    ],
    'H2O': [
        (298, 500,
         -203.6060,  1523.290,  -3196.413,  2474.455,
            3.855326, -256.5478,  -488.7163, -241.8264),
        (500, 1700,
           30.09200,    6.832514,    6.793435,  -2.534480,
            0.082139, -250.8810,   223.3967,  -241.8264)
    ],
}

# Standardbildungsenthalpien bei 298.15 K [kJ/mol]
Hf298 = {'CH4': -74.87, 'CO2': -393.51, 'CO': -110.53, 'H2': 0.0, 'H2O': -241.83}


def _get_coeffs(species: str, T: float):
    """Gibt die passenden Shomate-Koeffizienten für T [K] zurück."""
    for entry in SHOMATE[species]:
        T_min, T_max, *coeffs = entry
        if T_min <= T <= T_max:
            return coeffs
    raise ValueError(f"{species}: T={T} K außerhalb aller Gültigkeitsbereiche.")


def Cp(species: str, T: float) -> float:
    """Wärmekapazität [J/(mol·K)] bei T [K]."""
    A, B, C, D, E, F, G, H = _get_coeffs(species, T)
    t = T / 1000
    return A + B*t + C*t**2 + D*t**3 + E/t**2


def dH(species: str, T: float) -> float:
    """Enthalpiedifferenz H°(T) - H°(298.15) [kJ/mol]."""
    A, B, C, D, E, F, G, H = _get_coeffs(species, T)
    t = T / 1000
    return A*t + B*t**2/2 + C*t**3/3 + D*t**4/4 - E/t + F - H


def S(species: str, T: float) -> float:
    """Standardentropie S°(T) [J/(mol·K)]."""
    A, B, C, D, E, F, G, H = _get_coeffs(species, T)
    t = T / 1000
    return A*np.log(t) + B*t + C*t**2/2 + D*t**3/3 - E/(2*t**2) + G


def H_abs(species: str, T: float) -> float:
    """Absolute Standardenthalpie H°(T) = H°f,298 + dH [kJ/mol]."""
    return Hf298[species] + dH(species, T)
```

```python
T_arr = np.linspace(300, 1100, 400)

# Stöchiometrische Koeffizienten: positiv = Produkt, negativ = Edukt
nu = {'CH4': +1, 'H2O': +2, 'CO2': -1, 'H2': -4}

dH_rxn, dS_rxn, dG_rxn = [], [], []

for T in T_arr:
    try:
        dh = sum(n * H_abs(sp, T) for sp, n in nu.items())   # kJ/mol
        ds = sum(n * S(sp, T)     for sp, n in nu.items())   # J/(mol·K)
        dg = dh - T * ds / 1000                              # kJ/mol
        dH_rxn.append(dh)
        dS_rxn.append(ds)
        dG_rxn.append(dg)
    except ValueError:
        dH_rxn.append(np.nan)
        dS_rxn.append(np.nan)
        dG_rxn.append(np.nan)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle(r'CO$_2$ + 4 H$_2$ $\rightarrow$ CH$_4$ + 2 H$_2$O  –  Temperaturabhängigkeit', fontsize=13)

T_C = T_arr - 273.15

axes[0].plot(T_C, dH_rxn, color='#C04828')
axes[0].set_title(r'$\Delta H^\circ_{rxn}$ [kJ/mol]')
axes[0].set_xlabel('T [°C]'); axes[0].grid(alpha=0.3)
axes[0].axhline(0, color='k', lw=0.5)

axes[1].plot(T_C, dS_rxn, color='#185FA5')
axes[1].set_title(r'$\Delta S^\circ_{rxn}$ [J/(mol·K)]')
axes[1].set_xlabel('T [°C]'); axes[1].grid(alpha=0.3)
axes[1].axhline(0, color='k', lw=0.5)

axes[2].plot(T_C, dG_rxn, color='#0F6E56')
axes[2].set_title(r'$\Delta G^\circ_{rxn}$ [kJ/mol]')
axes[2].set_xlabel('T [°C]'); axes[2].grid(alpha=0.3)
axes[2].axhline(0, color='k', lw=0.8, ls='--', label=r'$\Delta G = 0$')
axes[2].legend(fontsize=9)

plt.tight_layout()
plt.show()
```

## Cp-Profile for all Spezies

```python
colors = {'CH4': '#C04828', 'CO2': "#3675B4", 'CO': "#238E73", 'H2': '#854F0B', 'H2O': "#3619A9"}
T_ranges = {'CH4': (300, 1300), 'CO2': (300, 1200), 'CO': (300, 1300), 'H2': (300, 2500), 'H2O': (300, 1700)}

fig, ax = plt.subplots(figsize=(9, 5))

for sp, (T_lo, T_hi) in T_ranges.items():
    T_s = np.linspace(T_lo, T_hi, 500)
    cp_vals = []
    for T in T_s:
        try:
            cp_vals.append(Cp(sp, T))
        except ValueError:
            cp_vals.append(np.nan)
    ax.plot(T_s - 273.15, cp_vals, label=sp, color=colors[sp], lw=1.8)

ax.set_xlabel('T [°C]', fontsize=11)
ax.set_ylabel(r'$C_p^\circ$ [J mol$^{-1}$ K$^{-1}$]', fontsize=11)
ax.set_title('Isobare Wärmekapazitäten (Shomate)', fontsize=12)
ax.legend(fontsize=10)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
